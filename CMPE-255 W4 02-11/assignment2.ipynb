```python
import numpy as np

# Given 4x4 matrix A
A = np.array([
    [4, -2, 2, 1],
    [1, 1, 0, 1],
    [-2, 1, 3, -1],
    [1, 3, -1, 2]
]);
```

## 1) Calculating the inverse matrix of A (1 pts)

```python
A_inv = np.linalg.inv(A)

print("Inverse of A:")
print(A_inv)
# print(
#     '\n',
#     "A @ A_inv is identical to A's identity matrix?",
#     np.allclose(A @ A_inv, np.eye(4))
# )
```

## 2) Calculating Eigenvalues and Eigenvectors for the 4x4 Matrix (2 pts).

```python
λ, v = np.linalg.eig(A)

print('eigenvalues:')
print(λ)
print('\neigenvectors:')
print(v)
# print('\nAv is identical to λv?', np.allclose(A @ v, λ * v))
```

## 3) After having the Eigenvalues and Eigenvectors,
- Constructing the eigen matrix P from eigenvectors: `P = eigenvectors`
- Constructing the diagonal matrix D from eigenvalues: `D = np.diag(eigenvalues)`
- Calculating the inverse of the eigen matrix: `P_inv = np.linalg.inv(P)`

Verify if the diagonalized matrix `A_diagonalized` is identical to the original matrix `A` (2 pts).

```python
P = v
D = np.diag(λ)
P_inv = np.linalg.inv(P)
A_diagonalized = P @ D @ P_inv

print(A_diagonalized);
print('\nA is identical to A_diagonalized?', np.allclose(A, A_diagonalized))
```

## 4)  Explain why this result (a diagonalized matrix is identical to the original matrix) is important in linear algebra (1 pts).

This is important because it proves that the original matrix can be broken down into its fundamental components, the eigenvalues and eigenvectors. When a matrix is diagonalized, performing operations such as Principal Component Analysis is much easier, we simply use the diagonalized components instead of performing expensive matrix operations. Being able to reconstruct the original matrix from its diagonalized components proves that we have accounted for all the variance in the original data when we perform dimensionality reduction.

## 5)  Make a python code using SVD example to get the same A (AR) (4 pts)

  - Decide your own range for the sigma

```python
from scipy.linalg import svd as SVD

U, S, V = SVD(A)
sigma = np.zeros((4, 4))
# The matrix is very small. We use the full range of the sigma, which is 4, to get the best reconstruction possible.
sigma_range = 4

for i in range(sigma_range):
    sigma[i, i] = S[i]

AR = U @ sigma @ V

print('A:')
print(A)
print('\nAR:')
print(AR)
print('\nA is identical to AR?', np.allclose(A, AR))
```
