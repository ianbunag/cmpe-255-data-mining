```python
import pandas as pd
import matplotlib.pyplot as plt

grades = pd.read_csv("grade.csv")
```

## 1. Draw histograms for “Grades” by “Gender (Male and Female)” with title, xlabel, and ylabel. (5 pts)

```python
hist = grades.hist(column="grade", by="msex")

for hist_ax in hist.flatten():
    hist_ax.set_xlabel("Grade")
    hist_ax.set_ylabel("Number of Students")
plt.suptitle("Histogram of Grades by Gender", fontsize=16)
plt.tight_layout()
```

## 2. Draw a box plot of Grades by Gender (Male and Female) on the same figure with title, xlabel, and ylabel included. Explain the meaning of each box plot. (5 pts)

```python
bp = grades.boxplot(column="grade", by="msex")

bp.set_xlabel("Gender")
bp.set_ylabel("Grade")
plt.title("Box plot of Grades by Gender", fontsize=16)
plt.suptitle("")
```

### Explanation
The box plots show the same distribution of grades between males and female student groups. Their p50 (median) and p75 (third quartile) have a value of 4.0, indicating that most students obtained high scores. Their IQR is 1.0, indicating that the difference between the middle 50% of students' grades are miniscule. There are a few exceptional and low performing students, with grades as high as 5.0 and as low as 2.0. There are also some outliers, achieving a failing grade of 1.0.
