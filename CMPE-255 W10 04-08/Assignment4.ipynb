Find a dataset which has two variables (you will call x and y, but they can have any names). You should have a suspicion or intuition that variable y is linearly dependent on variable x in the dataset.

```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import seaborn as sns

# https://www.kaggle.com/datasets/abhishek14398/salary-dataset-simple-linear-regression?resource=download&select=Salary_dataset.csv
df = pd.read_csv('assignment4-salary-dataset.csv')

df
```

## 1.Create a scatter plot of the data (2 pts)
## 2.Draw the least-square regression line through the scatter plot (3 pts).

```python
plt.figure()
years_experience = df['YearsExperience']
salary = df['Salary']
m, b = np.polyfit(years_experience, salary, 1)
regression_line = [m * x + b for x in years_experience]

plt.scatter(years_experience, salary)
plt.plot(years_experience, regression_line, color='gold')
plt.xlabel('YearsExperience')
plt.ylabel('Salary ($)')
plt.title(f'Linear Regression Line, m = {m:.2f}, b = {b:.2f}')

plt.show()
```

## 3.Calculate and print the coefficient of determination (2 pts) and the correlation coefficient (3 pts).

```python
years_experience_list = years_experience.values.reshape(-1, 1)
salary_list = salary.values.reshape(-1, 1)

pr = LinearRegression()
pr.fit(years_experience_list, salary_list)

r2 = r2_score(salary_list, pr.predict(years_experience_list))

print(f'Coefficient of determination: {r2:.2f}')

plt.title('Correlation coefficient')
sns.heatmap(data=df.corr(), annot=True, vmin=-1, vmax=1, cmap='coolwarm')
plt.show()
```
