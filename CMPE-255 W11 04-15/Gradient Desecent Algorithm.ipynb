# Gradient Descent (Linear Predictive Funtion)

Gradient Descent is an iterative algorithm that is used to minimize a function by finding the optimal parameters. 



Before we begin,

    > What is a cost function?
    > What is a Prediction Function?
    

Problem at hand : Help find the Predictive Fucntion for the below dataset of home prices in Monroe

But which one is the best fit line? Gradient Descent helps us find this best fit line 

Most common Cost Fucnction : Mean Square Error (MSE)

What are these optimal parameters?
1. Cost function
2. Number of iterations
3. A learning rate

What will be the output?

>A Prediction Function/Best Fit Function

# Code

```python
import numpy as np
```

```python
def gradient_descent(x,y,i,l):
    m_curr = b_curr = 0
    iterations = i
    n = len(x)
    learning_rate = l
    
    for i in range(iterations):
        y_predicted = m_curr * x + b_curr
        cost = (1/n) * sum([val**2 for val in (y-y_predicted)])
        md = -(2/n)*sum(x*(y-y_predicted))
        bd = -(2/n)*sum(y-y_predicted)
        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd
        print ("m {}, b {}, cost {} iteration {}".format(m_curr,b_curr,cost, i))
```

```python
x = np.array([26,30,32,36,40])
y = np.array([5500,5650,6100,6800,7250])
i=100 #number of iterations
l=0.008 #learning rate
```

```python
gradient_descent(x,y,i,l)
```
